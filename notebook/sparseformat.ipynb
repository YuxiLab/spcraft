# Direct Methods for Sparse Linear Systems: The Multifrontal Method
#### *A guide through the paper [The Multifrontal Method for Sparse Matrix Solution: Theory and Practice](https://epubs.siam.org/doi/10.1137/1034004)*

This notebook provides a complete, step-by-step implementation of the **multifrontal method** for solving large sparse symmetric positive definite linear systems ($Ax = b$), as presented by **Joseph W. H. Liu (1992)**.

The multifrontal method is one of the most powerful direct solver strategies in sparse linear algebra. It reinterprets the sparse Cholesky factorization ($A = LL^T$) as a sequence of partial factorizations on **smaller, dense frontal matrices**, organized hierarchically using an **elimination tree**.

In this notebook, we will implement the entire pipeline from scratch using **NumPy** and **SciPy**:
1. Defining and visualizing a sparse symmetric positive definite matrix.
2. Building and analyzing the **Elimination Tree** ($T(A)$).
3. Implementing the **Extend-Add Operator** ($\oplus$) for dense matrix superposition.
4. Implementing the **Stack-Based Multifrontal Cholesky Factorization** algorithm.
5. Verifying the result against standard dense and sparse libraries.
